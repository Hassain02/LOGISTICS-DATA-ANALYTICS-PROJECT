```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


df = pd.read_excel( r"C:\Users\LENOVO\Downloads\Unclean_Logistics_Dataset_12000plus_Rows.xlsx")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())


target = "Actual_Delivery_Days"

if target not in df.columns:
    raise ValueError(
        "The column 'Actual_Delivery_Days' was not found in the dataset."
    )

df[target] = pd.to_numeric(
    df[target],
    errors="coerce"
)

df = df.dropna(
    subset=[target]
).copy()

print("\nRows available for modeling:", len(df))


features = [
    "Distance_km",
    "Shipment_Weight_kg",
    "Quantity",
    "Transport_Mode",
    "Carrier",
    "Warehouse_ID",
    "Product_Category",
    "Promised_Delivery_Days",
    "Warehouse_Processing_Hours",
    "Temperature_C"
]


missing_features = [
    col for col in features
    if col not in df.columns
]

if missing_features:
    raise ValueError(
        "These required columns are missing: "
        + str(missing_features)
    )


X = df[features].copy()
y = df[target].copy()


categorical_features = [
    "Transport_Mode",
    "Carrier",
    "Warehouse_ID",
    "Product_Category"
]

numerical_features = [
    "Distance_km",
    "Shipment_Weight_kg",
    "Quantity",
    "Promised_Delivery_Days",
    "Warehouse_Processing_Hours",
    "Temperature_C"
]


for col in numerical_features:

    X[col] = pd.to_numeric(
        X[col],
        errors="coerce"
    )


for col in categorical_features:

    X[col] = X[col].astype(object)

    X[col] = X[col].where(
        pd.notna(X[col]),
        np.nan
    )

    X[col] = X[col].apply(
        lambda value:
        np.nan
        if pd.isna(value)
        else str(value).strip()
    )


print("\nMissing values in modeling data:")

display(
    X.isna().sum()
)


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Rows:", len(X_train))
print("Testing Rows:", len(X_test))


numerical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        )
    ]
)


categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)


linear_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LinearRegression()
        )
    ]
)


print("\nTraining Linear Regression...")

linear_model.fit(
    X_train,
    y_train
)

linear_predictions = linear_model.predict(
    X_test
)


linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)


print("\nLinear Regression Results")
print("-------------------------")
print("MAE :", round(linear_mae, 3))
print("RMSE:", round(linear_rmse, 3))
print("R2  :", round(linear_r2, 3))


random_forest = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=18,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


print("\nTraining Random Forest...")

random_forest.fit(
    X_train,
    y_train
)

rf_predictions = random_forest.predict(
    X_test
)


rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)


print("\nRandom Forest Results")
print("---------------------")
print("MAE :", round(rf_mae, 3))
print("RMSE:", round(rf_rmse, 3))
print("R2  :", round(rf_r2, 3))


model_results = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Random Forest"
    ],

    "MAE": [
        linear_mae,
        rf_mae
    ],

    "RMSE": [
        linear_rmse,
        rf_rmse
    ],

    "R2": [
        linear_r2,
        rf_r2
    ]

})


model_results = model_results.round(3)


print("\nModel Comparison")

display(
    model_results
)


best_model_index = model_results["RMSE"].idxmin()

best_model_name = model_results.loc[
    best_model_index,
    "Model"
]

print(
    "\nBest Model:",
    best_model_name
)


if best_model_name == "Linear Regression":

    best_predictions = linear_predictions

else:

    best_predictions = rf_predictions


plt.figure(
    figsize=(8, 5)
)

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.3
)

plt.xlabel(
    "Actual Delivery Days"
)

plt.ylabel(
    "Predicted Delivery Days"
)

plt.title(
    "Actual vs Predicted Delivery Days"
)

plt.grid(True)

plt.show()


prediction_df = X_test.copy()

prediction_df["Actual_Delivery_Days"] = (
    y_test.values
)

prediction_df["Predicted_Delivery_Days"] = (
    best_predictions
)

prediction_df["Prediction_Error"] = (
    prediction_df["Actual_Delivery_Days"]
    - prediction_df["Predicted_Delivery_Days"]
)


prediction_df["High_Delivery_Risk"] = (

    prediction_df["Predicted_Delivery_Days"]

    > prediction_df["Promised_Delivery_Days"]

).astype(int)


print(
    "\nTop 10 Predicted Longest Deliveries"
)

display(
    prediction_df
    .sort_values(
        "Predicted_Delivery_Days",
        ascending=False
    )
    .head(10)
)


high_risk_count = prediction_df[
    "High_Delivery_Risk"
].sum()


print(
    "\nHigh-risk shipments:",
    high_risk_count
)


if "Late_Delivery" not in df.columns:

    df["Delivery_Delay_Days"] = (
        df["Actual_Delivery_Days"]
        - df["Promised_Delivery_Days"]
    )

    df["Late_Delivery"] = (
        df["Delivery_Delay_Days"] > 0
    ).astype(int)


carrier_optimization = df.groupby(
    "Carrier"
).agg(

    Shipments=(
        "Shipment_ID",
        "count"
    ),

    Average_Delivery_Days=(
        "Actual_Delivery_Days",
        "mean"
    ),

    On_Time_Rate=(
        "Late_Delivery",
        lambda x: (1 - x.mean()) * 100
    ),

    Average_Cost=(
        "Shipping_Cost_INR",
        "mean"
    )

).round(2)


print(
    "\nCarrier Optimization Analysis"
)

display(
    carrier_optimization
    .sort_values(
        "On_Time_Rate",
        ascending=False
    )
)


df["Cost_per_km"] = (

    df["Shipping_Cost_INR"]
    / df["Distance_km"]

)


high_cost_shipments = (

    df[
        [
            "Shipment_ID",
            "Distance_km",
            "Shipping_Cost_INR",
            "Cost_per_km",
            "Carrier",
            "Transport_Mode"
        ]
    ]

    .sort_values(
        "Cost_per_km",
        ascending=False
    )

    .head(10)

)


print(
    "\nTop 10 Highest Cost-per-km Shipments"
)

display(
    high_cost_shipments
)


weekly_monitoring = df.groupby(
    "Order_Week"
).agg(

    Shipments=(
        "Shipment_ID",
        "count"
    ),

    Average_Delivery_Days=(
        "Actual_Delivery_Days",
        "mean"
    ),

    On_Time_Rate=(
        "Late_Delivery",
        lambda x: (1 - x.mean()) * 100
    ),

    Average_Shipping_Cost=(
        "Shipping_Cost_INR",
        "mean"
    ),

    Average_Warehouse_Processing_Hours=(
        "Warehouse_Processing_Hours",
        "mean"
    ),

    Average_Customer_Rating=(
        "Customer_Rating",
        "mean"
    )

).round(2)


print(
    "\nWeekly Monitoring"
)

display(
    weekly_monitoring
)


plt.figure(
    figsize=(10, 5)
)

plt.plot(
    weekly_monitoring.index,
    weekly_monitoring["On_Time_Rate"],
    marker="o"
)

plt.xlabel(
    "Order Week"
)

plt.ylabel(
    "On-Time Delivery Rate (%)"
)

plt.title(
    "Weekly On-Time Delivery Rate"
)

plt.grid(True)

plt.show()


with pd.ExcelWriter(
    "Week4_Predictive_Modeling_Results.xlsx",
    engine="openpyxl"
) as writer:

    model_results.to_excel(
        writer,
        sheet_name="Model_Comparison",
        index=False
    )

    prediction_df.to_excel(
        writer,
        sheet_name="Predictions",
        index=False
    )

    carrier_optimization.to_excel(
        writer,
        sheet_name="Carrier_Optimization"
    )

    weekly_monitoring.to_excel(
        writer,
        sheet_name="Weekly_Monitoring"
    )

    high_cost_shipments.to_excel(
        writer,
        sheet_name="High_Cost_Shipments",
        index=False
    )


print("\nWeek 4 completed successfully!")

print(
    "Results file: "
    "Week4_Predictive_Modeling_Results.xlsx"
)
```
